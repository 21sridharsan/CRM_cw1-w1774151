```python
#2b

import yfinance as yf
import pandas as pd

def calculate_betas(market_ticker, stock_list, start_date, end_date):
    market = yf.download(market_ticker, start=start_date, end=end_date)
    market['daily_return'] = market['Close'].pct_change()
    market.dropna(inplace=True)
    market_std = market['daily_return'].std()

    betas_list = []
    stock_names = []

    for stock in stock_list:
        try:
            data = yf.download(stock, start=start_date, end=end_date)
            data['daily_return'] = data['Close'].pct_change()
            data.dropna(inplace=True)

            if len(data) < 2:
                continue

            combined = pd.concat(
                [market['daily_return'], data['daily_return']],
                axis=1,
                join='inner'
            )
            combined.columns = ['market_return', 'stock_return']

            if combined['stock_return'].std() == 0:
                continue

            corr = combined['stock_return'].corr(combined['market_return'])
            beta_val = corr * (combined['stock_return'].std() / market_std)

            betas_list.append(beta_val)
            stock_names.append(stock)
        except:
            continue

    beta_df = pd.DataFrame({
        'Stock_Name': stock_names,
        'Beta': betas_list
    })

    return beta_df

betas_df = calculate_betas('^GSPC', close_prices.columns.tolist(), start, end)

betas_df.head()
```

```python
#3a

import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score

beta_no_na = betas_df.dropna(subset=['Beta']).copy()

X = beta_no_na['Beta'].values.reshape(-1, 1)

k_values = [2, 3, 4, 5, 6, 7, 8, 9]
scores = []

for k in k_values:
    model = AgglomerativeClustering(n_clusters=k)
    labels = model.fit_predict(X)
    score = silhouette_score(X, labels)
    scores.append(score)

plt.figure(figsize=(8, 5))
plt.plot(k_values, scores, marker='o')
plt.title("Agglomerative Clustering Silhouette Score vs K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

best_k_agglo = k_values[int(np.argmax(scores))]
print("Best K for Agglomerative Clustering:", best_k_agglo)

final_agglo = AgglomerativeClustering(n_clusters=best_k_agglo)
final_labels = final_agglo.fit_predict(X)

beta_no_na['AggloCluster'] = final_labels

```

```python
#4b

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

betas_local = betas_df.copy()
vol_local = vol_df.copy()

if 'Ticker' not in betas_local.columns:
    if 'Stock_Name' in betas_local.columns:
        betas_local = betas_local.rename(columns={'Stock_Name': 'Ticker'})
    else:
        betas_local = betas_local.rename(columns={betas_local.columns[0]: 'Ticker'})

if 'Ticker' not in vol_local.columns:
    vol_local = vol_local.reset_index()

if 'Ticker' not in vol_local.columns:
    if 'index' in vol_local.columns:
        vol_local = vol_local.rename(columns={'index': 'Ticker'})

if 'AnnualVolatility' not in vol_local.columns:
    other_cols = [c for c in vol_local.columns if c != 'Ticker']
    if len(other_cols) == 1:
        vol_local = vol_local.rename(columns={other_cols[0]: 'AnnualVolatility'})

temp_df = pd.merge(betas_local, vol_local, on='Ticker', how='inner')

X = temp_df[['Beta', 'AnnualVolatility']].dropna().values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

k_values = range(1, 11)
inertia_values = []

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia_values.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(k_values, inertia_values, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("K-Means Elbow Method")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
#4c

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

betas_local = betas_df.copy()
vol_local = vol_df.copy()

if 'Ticker' not in betas_local.columns:
    if 'Stock_Name' in betas_local.columns:
        betas_local = betas_local.rename(columns={'Stock_Name': 'Ticker'})
    else:
        betas_local = betas_local.rename(columns={betas_local.columns[0]: 'Ticker'})

if 'Ticker' not in vol_local.columns:
    vol_local = vol_local.reset_index()

if 'Ticker' not in vol_local.columns and 'index' in vol_local.columns:
    vol_local = vol_local.rename(columns={'index': 'Ticker'})

if 'AnnualVolatility' not in vol_local.columns:
    other_cols = [c for c in vol_local.columns if c != 'Ticker']
    if len(other_cols) == 1:
        vol_local = vol_local.rename(columns={other_cols[0]: 'AnnualVolatility'})

temp_df = pd.merge(betas_local, vol_local, on='Ticker', how='inner')
cluster_df = temp_df.dropna(subset=['Beta', 'AnnualVolatility']).copy()

X = cluster_df[['Beta', 'AnnualVolatility']].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

kmeans_final = KMeans(n_clusters=3, random_state=42, n_init=10)
final_labels = kmeans_final.fit_predict(X_scaled)

cluster_df['KMeansCluster'] = final_labels

plt.figure(figsize=(8, 6))
scatter = plt.scatter(
    cluster_df['Beta'],
    cluster_df['AnnualVolatility'],
    c=cluster_df['KMeansCluster'],
    s=25,
    cmap='viridis'
)
plt.xlabel("Beta")
plt.ylabel("Annual Volatility")
plt.title("K-Means Clusters (K = 3) – Beta vs Annual Volatility")
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(*scatter.legend_elements(), title="Cluster")
plt.show()
```

```python
#5b

import pandas as pd
import matplotlib.pyplot as plt

avg_daily_return = daily_returns.mean().to_frame('AvgDailyReturn').reset_index()
avg_daily_return = avg_daily_return.rename(columns={'index': 'Ticker'})

cluster_returns = pd.merge(cluster_df, avg_daily_return, on='Ticker', how='left')

cluster_summary = cluster_returns.groupby('KMeansCluster')[['Beta', 'AnnualVolatility', 'AvgDailyReturn']].mean()
print(cluster_summary)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
cluster_summary['Beta'].plot(kind='bar')
plt.title('Average Beta per Cluster')
plt.xlabel('Cluster')
plt.ylabel('Beta')
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.subplot(1, 3, 2)
cluster_summary['AnnualVolatility'].plot(kind='bar')
plt.title('Average Annual Volatility per Cluster')
plt.xlabel('Cluster')
plt.ylabel('Annual Volatility')
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.subplot(1, 3, 3)
cluster_summary['AvgDailyReturn'].plot(kind='bar')
plt.title('Average Daily Return per Cluster')
plt.xlabel('Cluster')
plt.ylabel('Avg Daily Return')
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
